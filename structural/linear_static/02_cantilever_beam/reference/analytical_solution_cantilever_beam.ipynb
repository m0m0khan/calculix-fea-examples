# Analytical Solution: Cantilever Beam

This notebook calculates two analytical results for the cantilever beam:

1. Maximum axial bending stress at the fixed end
2. Transverse displacement at the free end

The beam is treated using linear elastic Euler-Bernoulli beam theory.

A consistent **mm, s, N, tonne** unit system is used throughout the calculation.

## Problem Data

The cantilever beam has the following properties:

- Beam length: $L = 8000\,\mathrm{mm}$
- Square cross-section: $b = h = 1000\,\mathrm{mm}$
- Applied transverse force: $P = 9{,}000{,}000\,\mathrm{N}$
- Young's modulus: $E = 210{,}000\,\mathrm{N/mm^2}$

Because

$$
1\,\mathrm{N/mm^2}=1\,\mathrm{MPa},
$$

the Young's modulus can also be written as

$$
E=210{,}000\,\mathrm{MPa}.
$$

For a rectangular cross-section, the second moment of area is

$$
I=\frac{bh^3}{12}.
$$

The distance from the neutral axis to the outer surface is

$$
c=\frac{h}{2}.
$$

```python
# Consistent unit system: mm, s, N, tonne

P = 9_000_000.0    # Applied transverse force [N]
L = 8000.0         # Beam length [mm]
b = 1000.0         # Cross-section width [mm]
h = 1000.0         # Cross-section height [mm]

# 1 N/mm^2 = 1 MPa
E = 210_000.0      # Young's modulus [N/mm^2]

# Cross-sectional properties
I = b * h**3 / 12
c = h / 2

print(f"Second moment of area, I = {I:.6e} mm^4")
print(f"Outer-fiber distance, c = {c:.2f} mm")
```

## 1. Axial Bending Stress

For a cantilever beam subjected to a transverse force at its free end, the maximum bending moment occurs at the fixed end:

$$
M_{\max}=PL.
$$

Using force in newtons and length in millimetres, the bending moment has the unit

$$
[M_{\max}]=\mathrm{N\,mm}.
$$

The axial bending stress at a distance $y$ from the neutral axis is

$$
\sigma_{zz}(y)=-\frac{M_{\max}y}{I}.
$$

At the outer surfaces, $y=\pm c$. Therefore, the maximum bending-stress magnitude is

$$
\left|\sigma_{zz,\max}\right|=\frac{M_{\max}c}{I}.
$$

The resulting stress is expressed in $\mathrm{N/mm^2}$, which is equivalent to MPa.

```python
# Maximum bending moment at the fixed end
M_max = P * L

# Maximum axial bending stress at the outer surfaces
sigma_max = M_max * c / I

print(f"Maximum bending moment = {M_max:.6e} N mm")
print(f"Maximum bending moment = {M_max / 1e9:.2f} x 10^9 N mm")
print(f"Maximum axial bending-stress magnitude = {sigma_max:.2f} N/mm^2")
print(f"Maximum axial bending-stress magnitude = {sigma_max:.2f} MPa")
print(f"Outer-fiber stresses = +{sigma_max:.2f} MPa and -{sigma_max:.2f} MPa")
```

The analytical axial bending stresses at the fixed-end outer surfaces are

$$
\boxed{\sigma_{zz}=\pm432\,\mathrm{N/mm^2}}
$$

or, equivalently,

$$
\boxed{\sigma_{zz}=\pm432\,\mathrm{MPa}}.
$$

One outer surface of the beam is in tension, while the opposite outer surface is in compression.

The analytical value represents the nominal axial bending stress. A finite-element model may show a somewhat higher peak stress directly at the fully constrained face because of local three-dimensional constraint effects and stress extrapolation.

The finite-element maximum stress of approximately $465\,\mathrm{MPa}$ is about $7.6\%$ higher than the nominal analytical value of $432\,\mathrm{MPa}$.

## 2. Displacement at the Free End

According to Euler-Bernoulli beam theory, the transverse displacement at the free end of a cantilever beam subjected to an end force is

$$
u_y(L)=\frac{PL^3}{3EI}.
$$

Using the current unit system:

- $P$ is in N
- $L$ is in mm
- $E$ is in $\mathrm{N/mm^2}$
- $I$ is in $\mathrm{mm^4}$

Therefore, the calculated displacement is obtained directly in millimetres:

$$
[u_y]=\frac{\mathrm{N}\,\mathrm{mm^3}}{\left(\mathrm{N/mm^2}\right)\mathrm{mm^4}}=\mathrm{mm}.
$$

```python
# Transverse displacement at the free end
u_tip = P * L**3 / (3 * E * I)

print(f"Free-end transverse displacement = {u_tip:.4f} mm")
```

The analytical transverse displacement at the free end is

$$
\boxed{u_y(L)=87.77\,\mathrm{mm}}.
$$

In the CalculiX model:

- The beam axis is the global $z$-direction.
- The applied force acts in the global $y$-direction.
- The analytical bending displacement must be compared with the free-end displacement component $U_2$.
- The axial displacement along the beam axis is $U_3$, not $U_2$.

The finite-element displacement should therefore be checked using $U_2$ at nodes located exactly on the free-end face at

$$
z=8000\,\mathrm{mm}.
$$

## Summary

Using the consistent **mm, s, N, tonne** unit system, the analytical reference results are:

### Maximum axial bending stress

$$
\boxed{\left|\sigma_{zz,\max}\right|=432\,\mathrm{N/mm^2}=432\,\mathrm{MPa}}
$$

### Transverse displacement at the free end

$$
\boxed{u_y(L)=87.77\,\mathrm{mm}}
$$

The axial bending stress corresponds to the CalculiX stress component $S_{33}$ because the beam axis is the global $z$-direction.

The free-end bending displacement corresponds to the CalculiX displacement component $U_2$ because the applied force acts in the global $y$-direction.
